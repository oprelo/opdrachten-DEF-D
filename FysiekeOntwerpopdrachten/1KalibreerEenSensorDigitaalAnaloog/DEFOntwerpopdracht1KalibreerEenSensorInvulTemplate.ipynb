## *Opdracht 6*: Meten en fitten, Leerdoel

```python
# Zet hier jouw data in een voor jou handig formaat
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit

waardes = np.array([214, 202, 210, 201, 199, 297, 307, 216, 207, 198, 199, 287, 253, 265])
gewicht = np.array([0, 47.83, 6.03, 53.86, 95.66, 458.2, 506.03, 228, 1.9, 79.6, 109.56, 195+228, 228+47.83, 228+2*47.83])
waarde_error = 10
gewicht_error = 0.1
```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.
nul = np.zeros(100)+214
x = np.linspace(0, 500, 100)

def fit_func(w,a,b):
    return a*np.arctan(b*(w-270)) + 250

var, cov = curve_fit(fit_func, gewicht, waardes, p0=(100,50), maxfev=10000)
fit_waardes = fit_func(x,*var)

plt.figure(figsize=(9,5))
plt.plot(gewicht, waardes, 'b.', label='gemeten waardes')
plt.errorbar(gewicht, waardes, xerr=gewicht_error, yerr=waarde_error, fmt='none', ecolor='red', capsize=5)
plt.plot(x, nul, 'k--', label='nulwaarde')
plt.plot(x, fit_waardes, 'g--', label=f'fit: a={var[0]:.0f}+/-{np.sqrt(cov[0,0]):.0f}, b={var[1]:.3f}+/-{np.sqrt(cov[1,1]):.3f}')
plt.legend()
plt.xlabel('Gewicht (g)')
plt.ylabel('Waarde')
plt.show()
```

De curve fit is niet heel erg goed, maar er is ook lastig een eenduidig verband in de data te vinden. Ook werkt curve fit veel minder goed op het moment dat je fit op meer dan 2 variabele, dus hebben we de horizontale en verticale verschuiving moeten schatten.

Op basis van de data en criteria concluderen we dat de sensor niet bruikbaar is voor de gekozen toepassing. De sensor kon niet nauwkeurig een verschil meten in stappen van 1 gram, hierdoor voldoet de sensor niet aan de 2e criteria.
